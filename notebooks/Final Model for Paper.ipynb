```python
# ==================================================================
# EEG dementia pipeline v7 (AD / FTD / CN)  -- SINGLE CELL
# ds004504 (eyes-closed rest) + ds006036 (eyes-open photic)
#
# v7 = v6 model/training (UNCHANGED) + the pre-submission checklist:
#  0. FROZEN DESIGN, ONE CLEAN RUN: CFG['seed']=7. Headline declared BEFORE the run:
#     'NN fusion head, rest+photic' (classical weight 0). All other numbers are secondary.
#  1. LABEL-PERMUTATION TEST: subject labels shuffled, folds kept. Classical (many perms -> p-value) and
#     NN (a few perms -> sanity check that accuracy falls to ~0.33).
#  2. LENGTH CONFOUND: recordings are truncated (centre window) to a fixed number of segments for
#     evaluation; classical features, NN subject scores and the LENGTH_ONLY probe are all recomputed.
#  3. PAIRED COMPARISONS: paired bootstrap on balanced-accuracy difference + exact McNemar between heads;
#     ABLATIONS of every v6 change (modality dropout, train+val, multi-scale conv, head weights, all together).
#  4. DEMOGRAPHICS: age / sex / MMSE by group, group-difference tests, demographics-only baseline,
#     model correctness vs age/sex.
#  5. LIMITATIONS statement printed + saved (limitations_statement.txt).
#
# Leakage rules unchanged: all splits by subject, every fit (incl. normalisation stats) uses training subjects
# only; the test fold is never touched. Run order puts the headline FIRST; the extra analyses come after it
# and never feed back into it.
# ==================================================================

# !pip install mne scipy scikit-learn matplotlib seaborn tensorflow -q
import os, gc, glob, time, random, warnings
warnings.filterwarnings('ignore')
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'
from types import SimpleNamespace

import numpy as np
import pandas as pd
import matplotlib
if os.environ.get('EEG_SMOKE', '0') == '1': matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import kruskal, mannwhitneyu, binomtest, chi2_contingency, fisher_exact
import mne
mne.set_log_level('WARNING')

from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import (balanced_accuracy_score, f1_score, roc_auc_score,
                             confusion_matrix, classification_report)

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, regularizers

# ------------------------------------------------------------------
# 0. CONFIG  -- edit paths here
# ------------------------------------------------------------------
SMOKE_TEST = os.environ.get('EEG_SMOKE', '0') == '1'   # True = tiny synthetic run

PATHS = dict(
    rest   = "/kaggle/input/datasets/shivasubrahmanya/eeg-dataset/EEG 1/EEG 1",   # eyes-closed rest
    photic = "/kaggle/input/datasets/shivasubrahmanya/eeg-dataset/EEG 2/EEG 2",  # ds006036
)
USE_PREPROCESSED = True
OUT_DIR    = "/kaggle/working" if os.path.isdir("/kaggle/working") else "./eeg_out"
os.makedirs(OUT_DIR, exist_ok=True)
CACHE_FILE = os.path.join(OUT_DIR, "eeg_two_cond_cache_v2.npz")      # same cache as v2..v6 (reused)
EXT_FILE   = os.path.join(OUT_DIR, "eeg_ext_node_feats_v4.npy")      # extra node features (cached)

CHANNELS = ['Fp1','Fp2','F7','F3','Fz','F4','F8','T3','C3','Cz','C4','T4',
            'T5','P3','Pz','P4','T6','O1','O2']
REGIONS = {'frontal':  ['Fp1','Fp2','F7','F3','Fz','F4','F8'],
           'central':  ['C3','Cz','C4'],
           'temporal': ['T3','T4','T5','T6'],
           'parietal': ['P3','Pz','P4'],
           'occipital':['O1','O2']}
GROUP_MAP   = {'A':'AD','F':'FTD','C':'CN','AD':'AD','FTD':'FTD','CN':'CN'}
CLASS_NAMES = ['AD', 'CN', 'FTD']
COND_NAMES  = ['rest', 'photic']
TARGET_SFREQ, SEG_SEC, OVERLAP, AMP_THRESH = 128, 4, 0.5, 150.0
N_TIMES  = SEG_SEC * TARGET_SFREQ        # 512
N_CH, N_CLASSES = 19, 3
BANDS = {'delta':(0.5,4), 'theta':(4,8), 'alpha':(8,13), 'beta':(13,30), 'gamma':(30,45)}
N_BANDS = 5
BASE_NODE_FEAT = 8                       # 5 rel. band powers + 3 ratios (in the cache)
N_NODE_FEAT = BASE_NODE_FEAT             # updated to 16 after the extra features are added
EMB_T, EMB_G = 64, 64
MIN_SEGS = 20                            # recordings with fewer segments are ignored
TMP_KERNELS = (16, 32, 64)               # v6: multi-scale temporal kernels
TMP_F_PER_K = 8                          # filters per kernel -> 24 temporal filters

CFG = dict(
    seed=7,                              # v7: FINAL clean-run seed (v2-v6 used 42). Do not change after seeing results.
    n_folds=5, val_frac=0.2,
    n_seeds=5,                           # NN seeds averaged per fold
    rep_cv=5,                            # repeats of 5-fold CV for classical models
    ft_epochs=60, patience=12, sup_segs_per_rec=60, sup_bs=64,   # ft_epochs/patience only used if trainval=False
    lr=1e-3, wd=1e-3, label_smooth=0.05,
    head_w=(1.0, 0.7, 0.7),
    trainval=True,                       # train on train+val subjects, fixed epochs, SWA-style averaging
    fixed_epochs=25, swa_last=5,
    mod_drop=0.15,
    cls_w_sweep=[0.0, 0.25, 0.5], headline_w=0.0,
    cca_pca=20, cca_comp=5, cca_perm=100,
    dur_Ks=[1, 2, 4, 8, 16, 32, 64, 128], dur_reps=30, n_boot=1000,
    # ---- v7 additions ----
    run_pairs=True,
    run_lenmatch=True, len_match_pct=10,         # evaluation length = 10th percentile of recording lengths (in segments)
    run_demo=True,
    run_perm=True, clf_perm=200, nn_perm=5, perm_seeds=1,   # NN perms are a sanity check (too few for a p-value)
    run_ablations=True, abl_seeds=3,             # ablation runs use abl_seeds NN seeds (reference = first abl_seeds of main run)
)
HEADLINE_MODEL = 'NN fusion head, rest+photic'   # declared a priori (v7)
if SMOKE_TEST:
    CFG.update(n_folds=3, n_seeds=2, rep_cv=2, ft_epochs=2, fixed_epochs=3, swa_last=2, sup_segs_per_rec=20,
               cca_perm=10, dur_Ks=[1, 2, 4, 8], dur_reps=5, n_boot=50,
               clf_perm=10, nn_perm=1, abl_seeds=1)
    MIN_SEGS = 5

# model/training variant: the ablations override single entries of this dict
BASE_VAR = dict(trainval=CFG['trainval'], mod_drop=CFG['mod_drop'], head_w=CFG['head_w'],
                tmp_kernels=TMP_KERNELS, tmp_f_per_k=TMP_F_PER_K)

def set_seeds(s):
    random.seed(s); np.random.seed(s); tf.random.set_seed(s)

for g in tf.config.list_physical_devices('GPU'):
    try: tf.config.experimental.set_memory_growth(g, True)
    except Exception: pass
keras.mixed_precision.set_global_policy('float32')
print("TF", tf.__version__, "| GPUs:", tf.config.list_physical_devices('GPU'))
set_seeds(CFG['seed'])

# ------------------------------------------------------------------
# 1. DATA LOADING  (unchanged; cache is reused)
# ------------------------------------------------------------------
def load_labels(root):
    df = pd.read_csv(os.path.join(root, 'participants.tsv'), sep='\t')
    df.columns = [c.strip().lower() for c in df.columns]
    gcol = next(c for c in df.columns if c in ['group', 'diagnosis', 'label', 'class'])
    out = {}
    for _, r in df.iterrows():
        lab = GROUP_MAP.get(str(r[gcol]).strip().upper())
        if lab: out[str(r['participant_id']).strip()] = lab
    return out

def find_set_files(root, sub, keywords=None):
    for pat in [os.path.join(root, 'derivatives', 'eeglab', sub, 'eeg', '*.set'),
                os.path.join(root, 'derivatives', sub, 'eeg', '*.set'),
                os.path.join(root, sub, 'eeg', '*.set')]:
        hits = sorted(glob.glob(pat))
        if hits:
            if keywords:
                kh = [h for h in hits if any(k in os.path.basename(h).lower() for k in keywords)]
                hits = kh or hits
            return hits, 'derivatives' in pat
    return [], False

def load_recording(path, is_prep):
    raw = mne.io.read_raw_eeglab(path, preload=True, verbose=False)
    cmap = {c.upper(): c for c in raw.ch_names}
    keep = [cmap[c.upper()] for c in CHANNELS if c.upper() in cmap]
    assert len(keep) == N_CH, f"{path}: only {len(keep)}/19 channels"
    raw.pick_channels(keep, ordered=False); raw.reorder_channels(keep)
    if not (USE_PREPROCESSED and is_prep):
        raw.notch_filter(50.0, verbose=False); raw.filter(0.5, 45.0, verbose=False)
        raw.set_eeg_reference('average', projection=False, verbose=False)
    if raw.info['sfreq'] != TARGET_SFREQ: raw.resample(TARGET_SFREQ, verbose=False)
    return raw.get_data() * 1e6

def segment(data):
    step = int(N_TIMES * (1 - OVERLAP)); segs = []
    for s in range(0, data.shape[1] - N_TIMES + 1, step):
        w = data[:, s:s + N_TIMES]
        if np.max(np.abs(w)) <= AMP_THRESH: segs.append(w.astype(np.float32))
    if not segs: return None
    segs = np.stack(segs)
    mu = segs.mean(axis=(0, 2), keepdims=True); sd = segs.std(axis=(0, 2), keepdims=True) + 1e-6
    return ((segs - mu) / sd).astype(np.float32)                  # per-RECORDING z-score (label-free)

WIN, HOP = 128, 32
FREQS = np.fft.rfftfreq(WIN, 1.0 / TARGET_SFREQ)
BAND_MASKS = [(FREQS >= lo) & (FREQS < hi) for lo, hi in BANDS.values()]
HANN = np.hanning(WIN).astype(np.float32)

def spec_graph_features(segs, chunk=64):
    n = len(segs)
    nodes = np.zeros((n, N_CH, BASE_NODE_FEAT), np.float32)
    coh = np.zeros((n, N_BANDS, N_CH, N_CH), np.float16)
    for s in range(0, n, chunk):
        w = np.lib.stride_tricks.sliding_window_view(segs[s:s+chunk], WIN, axis=-1)[..., ::HOP, :]
        w = w - w.mean(-1, keepdims=True)
        X = np.fft.rfft(w * HANN, axis=-1)
        P = (np.abs(X) ** 2).mean(2)
        S = np.einsum('bcwf,bdwf->bcdf', X, X.conj(), optimize=True) / X.shape[2]
        C2 = np.abs(S) ** 2 / (P[:, :, None, :] * P[:, None, :, :] + 1e-12)
        bp = np.stack([P[..., m].sum(-1) for m in BAND_MASKS], -1) + 1e-12
        rel = bp / bp.sum(-1, keepdims=True)
        d, t, a, be = bp[..., 0], bp[..., 1], bp[..., 2], bp[..., 3]
        nodes[s:s+chunk] = np.concatenate(
            [np.log(rel), np.log(t / a)[..., None],
             np.log((d + t) / (a + be))[..., None], np.log(a / be)[..., None]], -1)
        coh[s:s+chunk] = np.stack([C2[..., m].mean(-1) for m in BAND_MASKS], 1)
    return nodes, coh

def graph_metrics(A):
    A = A.astype(np.float64).copy(); idx = np.arange(N_CH); A[:, idx, idx] = 0
    deg = A.sum(-1); dinv = 1 / np.sqrt(deg + 1e-8)
    Ls = np.eye(N_CH)[None] - dinv[:, :, None] * A * dinv[:, None, :]
    ev = np.clip(np.linalg.eigvalsh(Ls), 0, None); p = ev / (ev.sum(1, keepdims=True) + 1e-12)
    vne = -(p * np.log(p + 1e-12)).sum(1)
    iu = np.triu_indices(N_CH, 1); tri = A[:, iu[0], iu[1]]
    pe = tri / (tri.sum(1, keepdims=True) + 1e-12); edge_ent = -(pe * np.log(pe + 1e-12)).sum(1)
    ps = deg / (deg.sum(1, keepdims=True) + 1e-12); str_ent = -(ps * np.log(ps + 1e-12)).sum(1)
    Dm = 1 / (A + 1e-6); Dm[:, idx, idx] = 0
    for k in range(N_CH): Dm = np.minimum(Dm, Dm[:, :, k:k+1] + Dm[:, k:k+1, :])
    inv = 1 / (Dm + np.eye(N_CH)[None]); inv[:, idx, idx] = 0
    geff = inv.sum((1, 2)) / (N_CH * (N_CH - 1))
    W3 = np.cbrt(A / (A.max((1, 2), keepdims=True) + 1e-12))
    tr = np.diagonal(W3 @ W3 @ W3, axis1=1, axis2=2)
    ki = (A > 0).sum(-1); clus = (tr / (ki * (ki - 1) + 1e-12)).mean(1)
    return np.stack([vne, edge_ent, str_ent, tri.mean(1), tri.var(1), geff, clus], 1).astype(np.float32)
ENT_NAMES = ['vne', 'edge_entropy', 'strength_entropy', 'mean_coh', 'var_coh', 'global_eff', 'clustering']

def _synthetic(n_per_class=8, n_seg=30):
    rng = np.random.default_rng(0); recs, sub_lab = [], {}
    t = np.arange(N_TIMES) / TARGET_SFREQ; k = 0
    for ci, cn in enumerate(CLASS_NAMES):
        for _ in range(n_per_class):
            k += 1; sid = f"sub-{k:03d}"; sub_lab[sid] = cn
            for c in (0, 1):
                f0 = 7.0 + 2.5 * ci + c
                segs = rng.normal(size=(n_seg, N_CH, N_TIMES)) + \
                       1.0 * np.sin(2*np.pi*f0*t + rng.uniform(0, 6.3, (n_seg, N_CH, 1)))
                segs = ((segs - segs.mean((0, 2), keepdims=True)) / segs.std((0, 2), keepdims=True))
                recs.append((sid, c, segs.astype(np.float32)))
    return recs, sub_lab

def build_dataset():
    if os.path.exists(CACHE_FILE) and not SMOKE_TEST:
        print("Loading cache", CACHE_FILE); z = np.load(CACHE_FILE, allow_pickle=True)
        return SimpleNamespace(**{k: z[k] for k in z.files})
    if SMOKE_TEST:
        recs, sub_lab = _synthetic()
    else:
        lab_r, lab_p = load_labels(PATHS['rest']), load_labels(PATHS['photic'])
        clash = [s for s in set(lab_r) & set(lab_p) if lab_r[s] != lab_p[s]]
        assert not clash, f"Label mismatch between datasets for {clash[:5]}"
        sub_lab = {**lab_p, **lab_r}
        recs = []
        for sid in sorted(sub_lab):
            for c, (key, kw) in enumerate([('rest', ['eyesclosed', 'closed']), ('photic', None)]):
                files, prep = find_set_files(PATHS[key], sid, kw)
                if not files: print(f"  - {sid} [{COND_NAMES[c]}] no file"); continue
                parts = []
                for f in files:
                    try:
                        s_ = segment(load_recording(f, prep))
                        if s_ is not None: parts.append(s_)
                    except Exception as e: print(f"  x {sid} {f}: {e}")
                if parts:
                    recs.append((sid, c, np.concatenate(parts)))
                    print(f"  ok {sid} [{sub_lab[sid]}] {COND_NAMES[c]}: {len(recs[-1][2])} segs")
    sub_ids = np.array(sorted({r[0] for r in recs})); sid2i = {s: i for i, s in enumerate(sub_ids)}
    subj_y = np.array([CLASS_NAMES.index(sub_lab[s]) for s in sub_ids], np.int32)
    Xr, Xn, A, y, subj, cond, tpos = [], [], [], [], [], [], []
    for sid, c, segs in sorted(recs, key=lambda r: (sid2i[r[0]], r[1])):
        nodes, coh = spec_graph_features(segs); n = len(segs)
        Xr.append(segs.transpose(0, 2, 1).astype(np.float16)); Xn.append(nodes)
        A.append(coh.astype(np.float32).mean(1).astype(np.float16))
        y += [subj_y[sid2i[sid]]] * n; subj += [sid2i[sid]] * n; cond += [c] * n
        tpos.append(np.linspace(0, 1, n, dtype=np.float32))
    d = dict(Xr=np.concatenate(Xr), Xn=np.concatenate(Xn), A=np.concatenate(A),
             y=np.array(y, np.int32), subj=np.array(subj, np.int32),
             cond=np.array(cond, np.int8), tpos=np.concatenate(tpos),
             sub_ids=sub_ids, subj_y=subj_y)
    if not SMOKE_TEST: np.savez_compressed(CACHE_FILE, **d); print("Saved cache", CACHE_FILE)
    return SimpleNamespace(**d)

D = build_dataset()
S = len(D.sub_ids); N = len(D.y)

# ---- extra node features computed from the cached raw segments (from v4) ----
SUB_MASKS = [(FREQS >= lo) & (FREQS < hi) for lo, hi in [(8, 10), (10, 13), (13, 20), (20, 30)]]
ALPHA_M = (FREQS >= 6) & (FREQS <= 13); ALPHA_F = FREQS[ALPHA_M]
SE_M = (FREQS >= 0.5) & (FREQS <= 45)
N_EXT = 8

def extended_node_feats(xr):
    """xr: (b, T, C) z-scored segments -> (b, C, 8): 4 log sub-band shares, spectral entropy,
    alpha centroid frequency, log Hjorth mobility, log Hjorth complexity."""
    x = xr.astype(np.float32).transpose(0, 2, 1)                              # (b, C, T)
    w = np.lib.stride_tricks.sliding_window_view(x, WIN, axis=-1)[..., ::HOP, :]
    w = w - w.mean(-1, keepdims=True)
    P = (np.abs(np.fft.rfft(w * HANN, axis=-1)) ** 2).mean(2)                # (b, C, F)
    tot = P[..., SE_M].sum(-1) + 1e-12
    sub = np.stack([P[..., m].sum(-1) for m in SUB_MASKS], -1) + 1e-12
    rel_sub = np.log(sub / tot[..., None])
    pn = P[..., SE_M] / tot[..., None]
    sent = -(pn * np.log(pn + 1e-12)).sum(-1) / np.log(SE_M.sum())
    paf = (P[..., ALPHA_M] * ALPHA_F).sum(-1) / (P[..., ALPHA_M].sum(-1) + 1e-12)
    d1 = np.diff(x, axis=-1); d2 = np.diff(d1, axis=-1)
    v0, v1, v2 = x.var(-1) + 1e-12, d1.var(-1) + 1e-12, d2.var(-1) + 1e-12
    mob = np.sqrt(v1 / v0); comp = np.sqrt(v2 / v1) / mob
    return np.concatenate([rel_sub, sent[..., None], paf[..., None],
                           np.log(mob)[..., None], np.log(comp)[..., None]], -1).astype(np.float32)

def get_ext():
    if os.path.exists(EXT_FILE) and not SMOKE_TEST:
        e = np.load(EXT_FILE)
        if len(e) == N: print("Loading extended node features", EXT_FILE); return e
    t0 = time.time(); print("Computing extended node features ...")
    e = np.concatenate([extended_node_feats(D.Xr[i:i + 64]) for i in range(0, N, 64)])
    e = np.nan_to_num(e, nan=0.0, posinf=0.0, neginf=0.0)
    if not SMOKE_TEST: np.save(EXT_FILE, e)
    print(f"  done in {time.time()-t0:.0f}s"); return e

if D.Xn.shape[-1] == BASE_NODE_FEAT:
    D.Xn = np.concatenate([D.Xn, get_ext()], -1).astype(np.float32)
N_NODE_FEAT = int(D.Xn.shape[-1])
print("node features per channel:", N_NODE_FEAT)

RAW_COUNTS = np.zeros((S, 2))                                        # before MIN_SEGS filtering
for s_, c_ in zip(D.subj, D.cond): RAW_COUNTS[s_, c_] += 1
D.groups = {}
n_dropped = 0
for s in range(S):
    for c in (0, 1):
        idx = np.where((D.subj == s) & (D.cond == c))[0]
        if len(idx) >= MIN_SEGS: D.groups[(s, c)] = idx
        elif len(idx): n_dropped += 1; print(f"  dropping {D.sub_ids[s]} [{COND_NAMES[c]}]: only {len(idx)} segs")
print(f"\n{S} subjects | {N} segments | rest {(D.cond==0).sum()} | photic {(D.cond==1).sum()} | dropped recordings: {n_dropped}")
for i, cn in enumerate(CLASS_NAMES): print(f"  {cn}: {(D.subj_y==i).sum()} subjects")
_keep = np.zeros(N, bool)
for _idx in D.groups.values(): _keep[_idx] = True

# ------------------------------------------------------------------
# 2. SUBJECT-LEVEL FOLDS
# ------------------------------------------------------------------
def make_folds(subj_y, n_splits, val_frac, seed):
    skf = StratifiedKFold(n_splits, shuffle=True, random_state=seed); folds = []
    for tv, te in skf.split(np.zeros(len(subj_y)), subj_y):
        tr, va = train_test_split(tv, test_size=val_frac, stratify=subj_y[tv], random_state=seed)
        assert not (set(tr) & set(va)) and not (set(tr) & set(te)) and not (set(va) & set(te))
        folds.append((np.sort(tr), np.sort(va), np.sort(te)))
    return folds
FOLDS = make_folds(D.subj_y, CFG['n_folds'], CFG['val_frac'], CFG['seed'])   # always built from TRUE labels (also for permutation tests)
for i, (tr, va, te) in enumerate(FOLDS):
    print(f"fold {i}: train {len(tr)}  val {len(va)}  test {len(te)}  "
          f"test class counts {np.bincount(D.subj_y[te], minlength=3)}")

def seg_of(subj_set): return np.where(np.isin(D.subj, list(subj_set)))[0]

def capped_sample(subj_set, cap, rng):
    sset = set(int(s) for s in subj_set); out = []
    for (s, c), idx in D.groups.items():
        if s in sset: out.append(idx if len(idx) <= cap else rng.choice(idx, cap, replace=False))
    return np.concatenate(out)

class Batcher:
    def __init__(self, nmu, nsd, seg_y): self.mu, self.sd, self.y = nmu, nsd, seg_y
    def get(self, b):
        return (D.Xr[b].astype(np.float32), ((D.Xn[b] - self.mu) / self.sd).astype(np.float32),
                D.A[b].astype(np.float32), np.eye(2, dtype=np.float32)[D.cond[b]], self.y[b].astype(np.int32))

# ------------------------------------------------------------------
# 3. MODELS  (architecture identical to v6; the `var` dict only switches the ablated pieces)
# ------------------------------------------------------------------
def build_temporal_encoder(var):
    """Multi-scale log-band-power CNN (ShallowConvNet-style). var['tmp_kernels'] = temporal kernels in samples."""
    ks, fpk = var['tmp_kernels'], var['tmp_f_per_k']; sp = 2 * fpk * len(ks)
    inp = layers.Input((N_TIMES, N_CH))
    x = layers.GaussianNoise(0.02)(inp)
    x = layers.Reshape((N_TIMES, N_CH, 1))(x)
    br = [layers.Conv2D(fpk, (k, 1), padding='same', use_bias=False)(x) for k in ks]
    x = layers.Concatenate()(br) if len(br) > 1 else br[0]
    x = layers.BatchNormalization()(x)
    x = layers.DepthwiseConv2D((1, N_CH), depth_multiplier=2, use_bias=False,
                               depthwise_constraint=keras.constraints.MaxNorm(1.0))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Lambda(tf.square)(x)
    x = layers.AveragePooling2D((64, 1), strides=(16, 1))(x)
    x = layers.Lambda(lambda t: tf.math.log(tf.clip_by_value(t, 1e-6, 1e4)))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Reshape((-1, sp))(x)
    m = layers.GlobalAveragePooling1D()(x)
    s = layers.Lambda(lambda t: tf.math.reduce_std(t, axis=1))(x)
    z = layers.Concatenate()([m, s])
    h = layers.Dense(EMB_T)(layers.Dropout(0.3)(z))
    h = layers.LayerNormalization()(h)
    return keras.Model(inp, h, name='temporal_encoder')

class GraphPrep(layers.Layer):
    def __init__(self, topk=8, **kw): super().__init__(**kw); self.topk = topk
    def call(self, A):
        A = tf.cast(A, tf.float32); eye = tf.eye(N_CH)[None]
        A = A * (1.0 - eye)
        thr = tf.math.top_k(A, k=self.topk)[0][..., -1:]
        A = A * tf.cast(A >= thr, tf.float32)
        A = tf.maximum(A, tf.transpose(A, [0, 2, 1])) + eye
        dinv = tf.math.rsqrt(tf.reduce_sum(A, -1) + 1e-8)
        return -(dinv[:, :, None] * A * dinv[:, None, :])
    def compute_output_shape(self, s): return s
    def get_config(self): return {**super().get_config(), 'topk': self.topk}

class ChebConv(layers.Layer):
    def __init__(self, units, K=3, **kw): super().__init__(**kw); self.units, self.K = units, K
    def build(self, input_shape):
        F = input_shape[0][-1]
        self.W = self.add_weight(name='W', shape=(self.K * F, self.units), initializer='glorot_uniform',
                                 regularizer=regularizers.l2(3e-4))
        self.b = self.add_weight(name='b', shape=(self.units,), initializer='zeros')
        super().build(input_shape)
    def call(self, inputs):
        H, L = inputs; cheb = [H, tf.matmul(L, H)]
        for _ in range(2, self.K): cheb.append(2 * tf.matmul(L, cheb[-1]) - cheb[-2])
        Z = tf.concat(cheb[:self.K], -1)
        return tf.nn.relu(tf.tensordot(Z, self.W, axes=1) + self.b)
    def get_config(self): return {**super().get_config(), 'units': self.units, 'K': self.K}

def cheb_block(H, Ls, units, K, drop):
    z = ChebConv(units, K)([H, Ls]); z = layers.LayerNormalization()(z); z = layers.Dropout(drop)(z)
    r = H if H.shape[-1] == units else layers.Dense(units, use_bias=False)(H)
    return layers.Add()([z, r])

def build_graph_encoder(K=3, topk=8):
    n_in = layers.Input((N_CH, N_NODE_FEAT)); a_in = layers.Input((N_CH, N_CH))
    Ls = GraphPrep(topk)(a_in)
    x = layers.GaussianNoise(0.05)(n_in)
    h1 = cheb_block(x, Ls, 32, K, 0.2)
    h2 = cheb_block(h1, Ls, 64, K, 0.3)
    pooled = layers.Concatenate()([layers.GlobalAveragePooling1D()(h2), layers.GlobalMaxPooling1D()(h2),
                                   layers.GlobalAveragePooling1D()(h1)])
    per_node = layers.Flatten()(layers.Dense(8, activation='relu')(h2))
    g = layers.Concatenate()([pooled, per_node])
    g = layers.Dense(EMB_G, activation='relu')(layers.Dropout(0.3)(layers.LayerNormalization()(g)))
    return keras.Model([n_in, a_in], g, name='graph_encoder')

class ModalityDrop(layers.Layer):
    """Training-only. Randomly zeroes the temporal OR the graph embedding (each with prob rate/2)
    before the fusion head. Inference: identity. rate=0 -> identity (used by the ablation)."""
    def __init__(self, rate=0.15, **kw): super().__init__(**kw); self.rate = rate
    def call(self, inputs, training=None):
        h, g = inputs
        if training is not True or self.rate <= 0: return [h, g]
        u = tf.random.uniform([tf.shape(h)[0], 1])
        mh = tf.cast(u >= self.rate / 2, tf.float32)
        mg = tf.cast(tf.logical_or(u < self.rate / 2, u >= self.rate), tf.float32)
        return [h * mh, g * mg]
    def compute_output_shape(self, s): return s
    def get_config(self): return {**super().get_config(), 'rate': self.rate}

def build_classifier(enc_t, enc_g, var):
    x_in = layers.Input((N_TIMES, N_CH)); n_in = layers.Input((N_CH, N_NODE_FEAT))
    a_in = layers.Input((N_CH, N_CH)); c_in = layers.Input((2,))
    h = enc_t(x_in); g = enc_g([n_in, a_in])
    def head(z, name):
        z = layers.Concatenate()([z, c_in]); z = layers.Dropout(0.5)(z)
        z = layers.Dense(64, activation='relu', kernel_regularizer=regularizers.l2(1e-3))(z)
        return layers.Dense(N_CLASSES, name=name)(layers.Dropout(0.3)(z))
    h_d, g_d = ModalityDrop(var['mod_drop'])([h, g])
    fus = layers.Concatenate()([h_d, g_d])
    return keras.Model([x_in, n_in, a_in, c_in],
                       [head(fus, 'fusion'), head(h, 'tmp'), head(g, 'gr'), h, g], name='classifier')

# ------------------------------------------------------------------
# 4. AUGMENTATION + SUPERVISED TRAINING
# ------------------------------------------------------------------
def aug_eeg(x):
    B = tf.shape(x)[0]; ms = N_TIMES // 10
    sh = tf.random.uniform([B], -ms, ms + 1, dtype=tf.int32)
    idx = (tf.range(N_TIMES)[None, :] + sh[:, None]) % N_TIMES
    x = tf.gather(x, idx, batch_dims=1)
    x = x * tf.exp(tf.random.uniform([B, 1, N_CH], -0.1, 0.1))
    x = x + tf.random.normal(tf.shape(x)) * tf.random.uniform([B, 1, 1], 0.02, 0.06)
    return x * tf.cast(tf.random.uniform([B, 1, N_CH]) > 0.05, tf.float32)

def aug_graph(nodes, A):
    nodes = nodes + tf.random.normal(tf.shape(nodes), stddev=0.05)
    nodes = nodes * tf.cast(tf.random.uniform(tf.shape(nodes)) > 0.03, tf.float32)
    m = tf.cast(tf.random.uniform(tf.shape(A)) > 0.08, tf.float32)
    return nodes, A * tf.minimum(m, tf.transpose(m, [0, 2, 1]))

def reg_loss(m): return tf.add_n(m.losses) if m.losses else 0.0

def safe_opt(opt, vars_):
    try: opt.build(vars_)
    except Exception: pass
    return opt

def softmax_np(z): e = np.exp(z - z.max(-1, keepdims=True)); return e / e.sum(-1, keepdims=True)

_SIG = [tf.TensorSpec([None, N_TIMES, N_CH], tf.float32), tf.TensorSpec([None, N_CH, N_NODE_FEAT], tf.float32),
        tf.TensorSpec([None, N_CH, N_CH], tf.float32), tf.TensorSpec([None, 2], tf.float32)]
_FWD = {}

def predict_all(clf, bt, idx, bs=512):
    if id(clf) not in _FWD:
        _FWD[id(clf)] = tf.function(lambda x, n, a, c: clf([x, n, a, c], training=False), input_signature=_SIG)
    fwd = _FWD[id(clf)]
    outs = [[] for _ in range(5)]
    for i in range(0, len(idx), bs):
        x, n, a, c, _ = bt.get(idx[i:i + bs]); o = fwd(x, n, a, c)
        for k in range(5): outs[k].append(o[k].numpy())
    outs = [np.concatenate(o) for o in outs]
    return dict(probs=np.stack([softmax_np(outs[k]) for k in range(3)], 1).astype(np.float32),
                h=outs[3].astype(np.float32), g=outs[4].astype(np.float32))

def train_supervised(clf, bt, tr_set, va_set, rng, var, subj_y, verbose=True):
    """var['trainval']=True : tr_set already contains train+val subjects; fixed epochs, cosine decay,
                              average of the weights of the last `swa_last` epochs. va_set is ignored.
       var['trainval']=False: v5 behaviour (early stopping on va_set)."""
    tv = var['trainval']; E = CFG['fixed_epochs'] if tv else CFG['ft_epochs']
    ytr = subj_y[list(tr_set)]; cw = len(ytr) / (N_CLASSES * np.maximum(np.bincount(ytr, minlength=3), 1))
    cw_t = tf.constant(cw, tf.float32); ls = CFG['label_smooth']; hw = var['head_w']
    va_idx = None if tv else capped_sample(va_set, 60, np.random.default_rng(0))
    vars_ = clf.trainable_variables

    def wce(logits, y, w):
        oh = tf.one_hot(y, N_CLASSES) * (1 - ls) + ls / N_CLASSES
        return tf.reduce_sum(tf.nn.softmax_cross_entropy_with_logits(oh, logits) * w) / tf.reduce_sum(w)

    spe = max(1, len(capped_sample(tr_set, CFG['sup_segs_per_rec'], rng)) // CFG['sup_bs'])
    sched = keras.optimizers.schedules.CosineDecay(CFG['lr'], E * spe, alpha=0.05)
    opt = safe_opt(keras.optimizers.AdamW(learning_rate=sched, weight_decay=CFG['wd']), vars_)

    @tf.function(reduce_retracing=True)
    def step(x, n, a, c, y):
        x = aug_eeg(x); n, a = aug_graph(n, a); w = tf.gather(cw_t, y)
        with tf.GradientTape() as tape:
            o = clf([x, n, a, c], training=True)
            loss = sum(h * wce(o[k], y, w) for k, h in enumerate(hw)) + reg_loss(clf)
        grads = tape.gradient(loss, vars_)
        pairs = [(g, v) for g, v in zip(grads, vars_) if g is not None]
        g_, _ = tf.clip_by_global_norm([p[0] for p in pairs], 1.0)
        opt.apply_gradients(zip(g_, [p[1] for p in pairs])); return loss

    def validate():
        # early-stopping loss = fusion NLL + 0.5 x temporal NLL + 0.5 x graph NLL
        P = predict_all(clf, bt, va_idx)['probs']; y = bt.y[va_idx]; w = cw[y]
        nll = lambda o: float(-(np.log(o[np.arange(len(y)), y] + 1e-7) * w).sum() / w.sum())
        return nll(P[:, 0]) + 0.5 * nll(P[:, 1]) + 0.5 * nll(P[:, 2]), balanced_accuracy_score(y, P[:, 0].argmax(1))

    best, best_w, bad = np.inf, None, 0
    swa_sum, swa_n = None, 0
    for ep in range(E):
        idx = rng.permutation(capped_sample(tr_set, CFG['sup_segs_per_rec'], rng)); tl = []
        for i in range(0, len(idx) - CFG['sup_bs'] + 1, CFG['sup_bs']):
            x, n, a, c, y = bt.get(idx[i:i + CFG['sup_bs']]); tl.append(float(step(x, n, a, c, y)))
        if tv:
            if ep >= E - CFG['swa_last']:
                w_ = clf.get_weights()
                swa_sum = w_ if swa_sum is None else [a_ + b_ for a_, b_ in zip(swa_sum, w_)]; swa_n += 1
            if verbose and (ep % 5 == 0 or ep == E - 1): print(f"    ep {ep+1:3d}  train {np.mean(tl):.3f}")
        else:
            vl, vb = validate()
            if verbose and (ep % 5 == 0 or ep == E - 1):
                print(f"    ep {ep+1:3d}  train {np.mean(tl):.3f}  val_loss(fus+.5tmp+.5gr) {vl:.3f}  val_balAcc(seg) {vb:.3f}")
            if vl < best - 1e-4: best, best_w, bad = vl, clf.get_weights(), 0
            else:
                bad += 1
                if bad >= CFG['patience']:
                    if verbose: print(f"    early stop @ep {ep+1}")
                    break
    if tv and swa_n: clf.set_weights([a_ / swa_n for a_ in swa_sum])
    elif best_w is not None: clf.set_weights(best_w)

# ------------------------------------------------------------------
# 5. CCA (SVD-based; fit on train subjects only; evaluated on unseen subjects)
# ------------------------------------------------------------------
def cca_eval(h_tr, g_tr, h_te, g_te, y_te, rng):
    nc = CFG['cca_comp']
    clean = lambda a: np.nan_to_num(np.asarray(a, np.float64), nan=0., posinf=0., neginf=0.)

    def prep(tr, te):
        tr, te = clean(tr), clean(te)
        keep = tr.std(0) > 1e-6
        if keep.sum() < 3: raise ValueError(f"embedding collapsed ({keep.sum()} live dims)")
        tr, te = tr[:, keep], te[:, keep]
        sc = StandardScaler().fit(tr)
        k = max(2, min(CFG['cca_pca'], tr.shape[1], tr.shape[0] - 1))
        p = PCA(k, whiten=True, random_state=0).fit(sc.transform(tr))
        return p.transform(sc.transform(tr)), p.transform(sc.transform(te))

    Htr, Hte = prep(h_tr, h_te); Gtr, Gte = prep(g_tr, g_te)
    C = Htr.T @ Gtr / (len(Htr) - 1)
    U, s, Vt = np.linalg.svd(C, full_matrices=False)
    nk = min(nc, len(s)); Wx, Wy = U[:, :nk], Vt.T[:, :nk]
    Ttr, Utr, Tte, Ute = Htr @ Wx, Gtr @ Wy, Hte @ Wx, Gte @ Wy

    def rc(a, b):
        r = [np.corrcoef(a[:, k], b[:, k])[0, 1] for k in range(nk)]
        return r + [np.nan] * (nc - nk)
    null = []
    for _ in range(CFG['cca_perm']):
        perm = np.arange(len(Ute))
        for c in np.unique(y_te):
            m = np.where(y_te == c)[0]; perm[m] = rng.permutation(m)
        null.append(rc(Tte, Ute[perm]))
    pc = {CLASS_NAMES[c]: float(np.corrcoef(Tte[y_te == c, 0], Ute[y_te == c, 0])[0, 1])
          for c in np.unique(y_te) if (y_te == c).sum() > 5}
    return dict(r_train=rc(Ttr, Utr), r_test=rc(Tte, Ute), r_null=np.nanmean(null, 0).tolist(), r_class=pc)

# ------------------------------------------------------------------
# 6. STAGE 4 (descriptive): entropy biomarkers at SUBJECT level
# ------------------------------------------------------------------
print("\n=== Entropy / topology metrics ===")
ENT = np.concatenate([graph_metrics(D.A[i:i+2048].astype(np.float32)) for i in range(0, N, 2048)])
ent_df = pd.DataFrame(ENT, columns=ENT_NAMES)
ent_df['subj'], ent_df['cond'], ent_df['tpos'] = D.subj, D.cond, D.tpos
ent_df['label'] = [CLASS_NAMES[i] for i in D.y]
ent_df = ent_df[_keep].reset_index(drop=True)
sub_ent = ent_df.groupby(['subj', 'cond', 'label'])[ENT_NAMES].mean().reset_index()

def bh(p):
    p = np.asarray(p, float); o = np.argsort(p); r = np.empty_like(o); r[o] = np.arange(1, len(p) + 1)
    q = np.minimum.accumulate((p[o] * len(p) / r[o])[::-1])[::-1]; out = np.empty_like(q); out[o] = np.minimum(q, 1); return out

stat_rows = []
for c in (0, 1):
    sub = sub_ent[sub_ent.cond == c]
    if sub.label.nunique() < 3: continue
    ps = []
    for m in ENT_NAMES:
        groups = [sub[sub.label == k][m].values for k in CLASS_NAMES]
        H, p = kruskal(*groups); ps.append(p); stat_rows.append(dict(cond=COND_NAMES[c], metric=m, H=H, p=p))
    q = bh(ps)
    for r, qq in zip(stat_rows[-len(ENT_NAMES):], q): r['q_BH'] = qq
stat_df = pd.DataFrame(stat_rows)
print("Kruskal-Wallis on SUBJECT means (n = subjects, BH-corrected):\n", stat_df.round(4).to_string(index=False))
pw = []
for c in (0, 1):
    sub = sub_ent[sub_ent.cond == c]
    for m in ENT_NAMES:
        for a, b in [('AD', 'CN'), ('FTD', 'CN'), ('AD', 'FTD')]:
            x, y_ = sub[sub.label == a][m].values, sub[sub.label == b][m].values
            if len(x) > 2 and len(y_) > 2: pw.append(dict(cond=COND_NAMES[c], metric=m, pair=f"{a} vs {b}", p=mannwhitneyu(x, y_, alternative='two-sided')[1]))
pw_df = pd.DataFrame(pw); pw_df['q_BH'] = bh(pw_df.p.values); print("\nSignificant pairwise (q<0.05):\n", pw_df[pw_df.q_BH < 0.05].round(4).to_string(index=False))
stat_df.to_csv(f"{OUT_DIR}/stage4_kruskal_subject_level.csv", index=False); pw_df.to_csv(f"{OUT_DIR}/stage4_pairwise.csv", index=False)

pal = {'AD': '#E74C3C', 'CN': '#2ECC71', 'FTD': '#3498DB'}
fig, axes = plt.subplots(2, len(ENT_NAMES), figsize=(3.2 * len(ENT_NAMES), 7))
for c in (0, 1):
    for ax, m in zip(axes[c], ENT_NAMES):
        sub = sub_ent[sub_ent.cond == c]
        if len(sub) == 0: continue
        sns.boxplot(data=sub, x='label', y=m, order=CLASS_NAMES, palette=pal, ax=ax, width=.5, showfliers=False)
        sns.stripplot(data=sub, x='label', y=m, order=CLASS_NAMES, color='k', size=3, alpha=.6, ax=ax)
        ax.set_title(f"{m}\n[{COND_NAMES[c]}]", fontsize=9); ax.set_xlabel('')
plt.suptitle('Stage 4 - connectivity biomarkers (one dot = one subject)'); plt.tight_layout()
plt.savefig(f"{OUT_DIR}/stage4_entropy_biomarkers.png", dpi=150); plt.show()

ent_df['bin'] = np.minimum((ent_df.tpos * 5).astype(int), 4)
traj = ent_df.groupby(['subj', 'cond', 'label', 'bin'])['edge_entropy'].mean().reset_index()
fig, axes = plt.subplots(1, 2, figsize=(11, 3.8), sharey=True)
for c in (0, 1):
    for k in CLASS_NAMES:
        t = traj[(traj.cond == c) & (traj.label == k)].groupby('bin')['edge_entropy'].agg(['mean', 'sem'])
        if len(t): axes[c].errorbar(t.index, t['mean'], yerr=t['sem'], marker='o', color=pal[k], label=k, capsize=3)
    axes[c].set_title(f"{COND_NAMES[c]}"); axes[c].set_xlabel('Recording fifth (early -> late)'); axes[c].grid(alpha=.3)
axes[0].set_ylabel('Edge entropy (mean +/- SEM over subjects)'); axes[0].legend()
plt.suptitle('Stage 6b - within-recording entropy trajectory'); plt.tight_layout()
plt.savefig(f"{OUT_DIR}/stage6_entropy_trajectory.png", dpi=150); plt.show()

# ------------------------------------------------------------------
# 7. CLASSICAL SUBJECT-LEVEL MODELS (repeated subject-level CV)
# ------------------------------------------------------------------
REG_IDX = {r: [CHANNELS.index(c) for c in chs] for r, chs in REGIONS.items()}
REG_NAMES = list(REGIONS)

def regional_feats(idx):
    nf = D.Xn[idx].mean(0)
    A_ = D.A[idx].astype(np.float32).mean(0)
    f = [nf[REG_IDX[r]].mean(0) for r in REG_NAMES]
    for i, r in enumerate(REG_NAMES):
        ii = REG_IDX[r]; iu = np.triu_indices(len(ii), 1)
        f.append(np.array([A_[np.ix_(ii, ii)][iu].mean()]))
        for r2 in REG_NAMES[i + 1:]: f.append(np.array([A_[np.ix_(ii, REG_IDX[r2])].mean()]))
    f.append(ENT[idx].mean(0))
    return np.concatenate([np.ravel(x) for x in f]).astype(np.float32)

def build_feature_matrices(groups):
    """groups: dict (subject, cond) -> segment indices. Used for full-length AND length-matched features."""
    nfeat = len(regional_feats(next(iter(groups.values()))))
    R = np.full((S, 2 * nfeat), np.nan, np.float32)
    F = N_CH * N_NODE_FEAT + len(ENT_NAMES); FULL = np.full((S, 2 * F), np.nan, np.float32)
    for (s, c), idx in groups.items():
        R[s, c*nfeat:(c+1)*nfeat] = regional_feats(idx)
        FULL[s, c*F:(c+1)*F] = np.concatenate([D.Xn[idx].mean(0).ravel(), ENT[idx].mean(0)])
    return R, FULL
X_REG, X_FULL = build_feature_matrices(D.groups)
X_LEN = np.log1p(RAW_COUNTS).astype(np.float32)                   # recording-length only (confound probe)
print(f"\nregional feature matrix {X_REG.shape} | full-feature matrix {X_FULL.shape}")

def _pipe(clf): return make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), clf)
K_SEL = min(40, X_REG.shape[1])
MODELS = {
    'LR_regional': lambda: _pipe(LogisticRegression(C=0.1, max_iter=5000, class_weight='balanced')),
    'LDA_shrink':  lambda: _pipe(LinearDiscriminantAnalysis(solver='lsqr', shrinkage='auto')),
    'SVM_rbf':     lambda: _pipe(SVC(C=1.0, kernel='rbf', probability=True, class_weight='balanced', random_state=0)),
    'LR_selected': lambda: make_pipeline(SimpleImputer(strategy='median'), StandardScaler(),
                                         SelectKBest(f_classif, k=K_SEL),
                                         LogisticRegression(C=0.2, max_iter=5000, class_weight='balanced')),
    'ExtraTrees':  lambda: make_pipeline(SimpleImputer(strategy='median'),
                                         ExtraTreesClassifier(n_estimators=400, min_samples_leaf=2, max_features='sqrt',
                                                              class_weight='balanced_subsample', n_jobs=-1, random_state=0)),
}
CLS_MEMBERS = ['LR_regional', 'LDA_shrink', 'LR_selected', 'ExtraTrees']
LR_FULL = lambda: _pipe(LogisticRegression(C=0.05, max_iter=5000, class_weight='balanced'))
LR_LEN  = lambda: _pipe(LogisticRegression(C=1.0, max_iter=2000, class_weight='balanced'))

def classical_cv(X, model_fn, y=None, reps=None):
    """Subject-level CV on the fixed FOLDS (r=0) + re-seeded folds (r>0). y: labels used for fit/score
    (default true labels; pass permuted labels for the permutation test; folds always come from true labels)."""
    y = D.subj_y if y is None else y; reps = CFG['rep_cv'] if reps is None else reps
    accs, P0 = [], None
    for r in range(reps):
        folds = FOLDS if r == 0 else make_folds(D.subj_y, CFG['n_folds'], CFG['val_frac'], CFG['seed'] + r)
        P = np.full((S, 3), np.nan)
        for tr, va, te in folds:
            trv = np.concatenate([tr, va]); m = model_fn(); m.fit(X[trv], y[trv])
            pr = m.predict_proba(X[te])
            P[np.ix_(te, m.classes_.astype(int))] = pr if pr.shape[1] == 3 else pr   # classes_ = subset of 0..2
            if pr.shape[1] < 3:
                miss = [c for c in range(3) if c not in m.classes_]; P[np.ix_(te, miss)] = 0.0
        accs.append(balanced_accuracy_score(y, P.argmax(1)))
        if r == 0: P0 = P
    return float(np.mean(accs)), float(np.std(accs)), P0

cls_rows, cls_P = [], {}
for name, fn in MODELS.items():
    m_, s_, P0 = classical_cv(X_REG, fn); cls_P[name] = P0
    cls_rows.append(dict(model=name, features='regional(both)', balAcc_mean=m_, balAcc_sd=s_, balAcc_fold0=balanced_accuracy_score(D.subj_y, P0.argmax(1))))
m_, s_, P0 = classical_cv(X_FULL, LR_FULL)
cls_rows.append(dict(model='LR_fullfeat', features='per-channel(both)', balAcc_mean=m_, balAcc_sd=s_, balAcc_fold0=balanced_accuracy_score(D.subj_y, P0.argmax(1))))
CLS_ENS = np.mean([cls_P[k] for k in CLS_MEMBERS], 0)
cls_rows.append(dict(model='CLASSICAL_ENSEMBLE (fold0 only)', features='regional(both)', balAcc_mean=np.nan, balAcc_sd=np.nan,
                     balAcc_fold0=balanced_accuracy_score(D.subj_y, CLS_ENS.argmax(1))))
m_, s_, P0 = classical_cv(X_LEN, LR_LEN)
cls_rows.append(dict(model='LENGTH_ONLY (confound probe)', features='log #segments rest/photic', balAcc_mean=m_, balAcc_sd=s_, balAcc_fold0=balanced_accuracy_score(D.subj_y, P0.argmax(1))))
cls_df = pd.DataFrame(cls_rows); cls_df.to_csv(f"{OUT_DIR}/classical_results.csv", index=False)
print("\n" + "=" * 78 + f"\nCLASSICAL MODELS  ({CFG['rep_cv']} x {CFG['n_folds']}-fold subject-level CV, chance = 0.333)\n" + "=" * 78)
print(cls_df.round(3).to_string(index=False))
print("If LENGTH_ONLY is well above 0.333, recording length (protocol) leaks group information -> see length-matched analysis (section 13).")

# ------------------------------------------------------------------
# 8. NN CROSS-VALIDATION (multi-seed) -- one reusable runner (main run, permutations, ablations)
# ------------------------------------------------------------------
def run_nn_cv(tag, subj_y, seg_y, var, n_seeds, do_cca=False, verbose=True):
    """5-fold subject-level CV on FOLDS. subj_y/seg_y are the labels used for TRAINING (true labels, or permuted).
    Returns per-seed out-of-fold probabilities (n_seeds, N, 3 heads, 3 classes), filled-mask, CCA results."""
    oof_s = np.zeros((n_seeds, N, 3, 3), np.float32); fl = np.zeros(N, bool); cca_ = []
    for fi, (tr, va, te) in enumerate(FOLDS):
        fit_set = np.concatenate([tr, va]) if var['trainval'] else tr
        assert not (set(fit_set.tolist()) & set(te.tolist()))               # test subjects never used for fitting
        fit_idx, te_idx = seg_of(fit_set), seg_of(te)
        assert not (set(D.subj[fit_idx]) & set(D.subj[te_idx]))
        fit_keep = fit_idx[_keep[fit_idx]]
        mu = D.Xn[fit_keep].mean((0, 1), keepdims=True)[0]; sd = D.Xn[fit_keep].std((0, 1), keepdims=True)[0] + 1e-6
        bt = Batcher(mu, sd, seg_y)                                          # stats from FIT subjects only
        for si in range(n_seeds):
            print(f"\n===== [{tag}] FOLD {fi+1}/{len(FOLDS)} | seed {si+1}/{n_seeds} =====") if verbose else \
                print(f"  [{tag}] fold {fi+1}/{len(FOLDS)} seed {si+1}/{n_seeds}")
            keras.backend.clear_session(); _FWD.clear(); gc.collect()
            sseed = CFG['seed'] + 100 * fi + si; set_seeds(sseed); rng = np.random.default_rng(sseed)
            enc_t, enc_g = build_temporal_encoder(var), build_graph_encoder()
            clf = build_classifier(enc_t, enc_g, var)
            if verbose and fi == 0 and si == 0:
                print(f"    params: temporal {enc_t.count_params():,} | graph {enc_g.count_params():,} | total {clf.count_params():,}")
            train_supervised(clf, bt, fit_set, va, rng, var, subj_y, verbose=verbose)
            pr = predict_all(clf, bt, te_idx); oof_s[si, te_idx] = pr['probs']
            if do_cca and si == 0:
                ptr = predict_all(clf, bt, capped_sample(fit_set, CFG['sup_segs_per_rec'], rng))
                print(f"    embeddings: h std {pr['h'].std():.3f} (dead {(pr['h'].std(0)<1e-6).sum()}/{pr['h'].shape[1]}) | "
                      f"g std {pr['g'].std():.3f} (dead {(pr['g'].std(0)<1e-6).sum()}/{pr['g'].shape[1]})")
                try:
                    cca_.append(cca_eval(ptr['h'], ptr['g'], pr['h'], pr['g'], seg_y[te_idx], rng))
                    r = cca_[-1]
                    print(f"    CCA r1: train {r['r_train'][0]:.3f} | TEST {r['r_test'][0]:.3f} | null {r['r_null'][0]:.3f}")
                except Exception as e:
                    print(f"    CCA skipped: {e}")
        fl[te_idx] = True
    return oof_s, fl, cca_

oof_s, filled, cca_res = run_nn_cv('MAIN', D.subj_y, D.y, BASE_VAR, CFG['n_seeds'], do_cca=True)
oof = oof_s.mean(0)                                                          # (N, 3 heads, 3 classes)
np.savez_compressed(f"{OUT_DIR}/oof_predictions.npz", filled=filled, oof=oof, oof_seeds=oof_s)

# ------------------------------------------------------------------
# 9. SUBJECT-LEVEL RESULTS (out-of-fold)  -- HEADLINE FIRST
# ------------------------------------------------------------------
def subject_scores(P, which, groups=None, fl=None):
    groups = D.groups if groups is None else groups; fl = filled if fl is None else fl
    conds = {'rest': (0,), 'photic': (1,), 'both': (0, 1)}[which]; out = np.full((S, 3), np.nan)
    for s in range(S):
        parts = [np.log(P[groups[(s, c)]] + 1e-6).mean(0) for c in conds
                 if (s, c) in groups and fl[D.groups[(s, c)][0]]]
        if parts: out[s] = np.mean(parts, 0)
    return out

def boot_ci(yt, yp, rng):
    v = [balanced_accuracy_score(yt[i], yp[i]) for i in (rng.integers(0, len(yt), len(yt)) for _ in range(CFG['n_boot']))]
    return np.percentile(v, [2.5, 97.5])

def summarize(yt, pp, rng):
    yp = pp.argmax(1); lo, hi_ = boot_ci(yt, yp, rng)
    try: auc = roc_auc_score(yt, pp / pp.sum(1, keepdims=True), multi_class='ovr', average='macro')
    except Exception: auc = np.nan
    return dict(bal_acc=balanced_accuracy_score(yt, yp), ci_lo=lo, ci_hi=hi_, macro_F1=f1_score(yt, yp, average='macro'), AUC=auc)

rng = np.random.default_rng(1); rows = []; store = {}
HEADS = ['fusion', 'tmp (temporal only)', 'gr (graph only)']
for hi_i, head in enumerate(HEADS):
    for which in ['rest', 'photic', 'both']:
        sc = subject_scores(oof[:, hi_i], which); ok = ~np.isnan(sc).any(1)
        yt, yp = D.subj_y[ok], sc[ok].argmax(1)
        rows.append(dict(head=head, input=which, n_subj=int(ok.sum()), **summarize(yt, softmax_np(sc[ok]), rng)))
        store[(hi_i, which)] = (yt, yp, sc, ok)
res = pd.DataFrame(rows); res.to_csv(f"{OUT_DIR}/subject_level_results.csv", index=False)
print("\n" + "=" * 78 + f"\nNN SUBJECT-LEVEL OUT-OF-FOLD RESULTS ({CFG['n_seeds']}-seed ensemble; chance = 0.333)\n" + "=" * 78)
print(res.round(3).to_string(index=False))

# ---- NN ensembles (+ optional classical blend; weight sweep reported openly) ----
_, _, sc_f, ok_f = store[(0, 'both')]; _, _, sc_t, ok_t = store[(1, 'both')]; _, _, sc_g, ok_g = store[(2, 'both')]
ok_e = ok_f & ok_g & ok_t; yt_e_all = D.subj_y[ok_e]
NN_CANDS = {
    HEADLINE_MODEL:    softmax_np(sc_f[ok_e]),                                       # declared headline (v7)
    'NN all-3 heads':  softmax_np((sc_f[ok_e] + sc_t[ok_e] + sc_g[ok_e]) / 3),
    'NN fusion+graph': softmax_np(0.5 * sc_f[ok_e] + 0.5 * sc_g[ok_e]),
}
ens_rows, ens_store = [], {}
for nm, p_nn in NN_CANDS.items():
    for w in CFG['cls_w_sweep']:
        pp = (1 - w) * p_nn + w * CLS_ENS[ok_e]
        is_head = (nm == HEADLINE_MODEL and abs(w - CFG['headline_w']) < 1e-9)
        ens_rows.append(dict(model=f"{nm} + {w:.2f} x classical" + ("   <- HEADLINE" if is_head else ""),
                             n_subj=int(ok_e.sum()), **summarize(yt_e_all, pp, rng)))
        ens_store[(nm, w)] = (yt_e_all, pp)
ens_rows.append(dict(model="Classical ensemble alone", n_subj=int(ok_e.sum()), **summarize(yt_e_all, CLS_ENS[ok_e], rng)))
ens_df = pd.DataFrame(ens_rows); ens_df.to_csv(f"{OUT_DIR}/ensemble_results.csv", index=False)
print("\n" + "=" * 78 + "\nENSEMBLES (same folds)\n" + "=" * 78)
print(ens_df.round(3).to_string(index=False))

yt_e, pp_e = ens_store[(HEADLINE_MODEL, CFG['headline_w'])]
hl = summarize(yt_e, pp_e, np.random.default_rng(11))
print("\n" + "#" * 78)
print(f"HEADLINE (declared a priori, seed {CFG['seed']}, single clean run): {HEADLINE_MODEL}")
print(f"  subject-level balanced accuracy = {hl['bal_acc']:.3f}  (95% bootstrap CI {hl['ci_lo']:.3f} - {hl['ci_hi']:.3f}) | "
      f"macro-F1 {hl['macro_F1']:.3f} | macro-AUC {hl['AUC']:.3f} | n = {int(ok_e.sum())} subjects | chance = 0.333")
print("  Report this number even if it is lower than v2-v6 numbers. Everything below is secondary / diagnostic.")
print("#" * 78)
print("Design choices were informed by v2-v6 on these same subjects (seed 42); v7 uses a new seed but NOT new subjects, "
      "so the result is still slightly optimistic. State this in the paper.")

fig, ax = plt.subplots(1, 1, figsize=(6.5, 4))
for j, which in enumerate(['rest', 'photic', 'both']):
    s_ = res[res['input'] == which]; xs = np.arange(len(s_)) + (j - 1) * 0.27
    ax.bar(xs, s_.bal_acc, 0.27, yerr=[s_.bal_acc - s_.ci_lo, s_.ci_hi - s_.bal_acc], capsize=3, label=which)
ax.axhline(1/3, color='gray', ls='--'); ax.set_xticks(range(3)); ax.set_xticklabels(['fusion', 'temporal', 'graph'])
ax.set_ylim(0, 1); ax.grid(axis='y', alpha=.3); ax.legend()
ax.set_ylabel('Subject-level balanced accuracy (95% bootstrap CI)')
plt.tight_layout(); plt.savefig(f"{OUT_DIR}/cv_results.png", dpi=150); plt.show()

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, (ttl, yt_, yp_) in zip(axes, [(f"HEADLINE: {HEADLINE_MODEL}", yt_e, pp_e.argmax(1)),
                                      ("NN all-3 heads (secondary)", *[ens_store[('NN all-3 heads', 0.0)][0],
                                                                       ens_store[('NN all-3 heads', 0.0)][1].argmax(1)])]):
    sns.heatmap(confusion_matrix(yt_, yp_), annot=True, fmt='d', cmap='Blues', xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, ax=ax)
    ax.set_title(ttl + "\n(subject counts, out-of-fold)"); ax.set_xlabel('Predicted'); ax.set_ylabel('True')
plt.tight_layout(); plt.savefig(f"{OUT_DIR}/confusion_subject_level.png", dpi=150); plt.show()
print("\n[HEADLINE] classification report:\n", classification_report(yt_e, pp_e.argmax(1), target_names=CLASS_NAMES, digits=3, zero_division=0))

# ------------------------------------------------------------------
# 10. CCA: out-of-sample temporal <-> spatial coupling (descriptive only)
# ------------------------------------------------------------------
if not cca_res:
    print("no CCA results (all folds skipped)")
else:
    R = {k: np.array([f[k] for f in cca_res], float) for k in ['r_train', 'r_test', 'r_null']}
    x = np.arange(1, CFG['cca_comp'] + 1)
    fig, ax = plt.subplots(1, 1, figsize=(6, 4))
    for j, (k, lab) in enumerate([('r_train', 'train subjects'), ('r_test', 'UNSEEN subjects'), ('r_null', 'null (within-class shuffle)')]):
        ax.bar(x + (j - 1) * 0.27, np.nanmean(R[k], 0), 0.27, yerr=np.nanstd(R[k], 0), capsize=3, label=lab)
    ax.set_title(f"CCA temporal<->graph  ({len(cca_res)}/{CFG['n_folds']} folds)"); ax.set_xlabel('Canonical component'); ax.grid(axis='y', alpha=.3)
    ax.set_ylabel('Correlation r'); ax.legend(fontsize=8)
    print(f"\ncanonical r (mean over folds)  train {np.round(np.nanmean(R['r_train'],0),3)}  "
          f"TEST {np.round(np.nanmean(R['r_test'],0),3)}  null {np.round(np.nanmean(R['r_null'],0),3)}")
    pcs = pd.DataFrame([f['r_class'] for f in cca_res]).mean().round(3).to_dict(); print(f"    component-1 test r per class: {pcs}")
    plt.tight_layout(); plt.savefig(f"{OUT_DIR}/stage5_cca_out_of_sample.png", dpi=150); plt.show()

# ------------------------------------------------------------------
# 11. RECORDING-DURATION ANALYSIS (cross-sectional: "how much EEG is needed",
#     NOT "how early can disease be detected")
# ------------------------------------------------------------------
def duration_curve(P, which, rng):
    conds = {'rest': (0,), 'photic': (1,), 'both': (0, 1)}[which]; Ks = CFG['dur_Ks']
    L = np.log(P + 1e-6); mean, lo, hi = [], [], []
    subs = [s for s in range(S) if any((s, c) in D.groups and filled[D.groups[(s, c)][0]] for c in conds)]
    yt = D.subj_y[subs]
    for K in Ks:
        accs = []
        for _ in range(CFG['dur_reps']):
            pred = []
            for s in subs:
                parts = []
                for c in conds:
                    idx = D.groups.get((s, c))
                    if idx is None: continue
                    k = min(K, len(idx)); st = rng.integers(0, len(idx) - k + 1); parts.append(L[idx[st:st + k]].mean(0))
                pred.append(np.mean(parts, 0).argmax())
            accs.append(balanced_accuracy_score(yt, pred))
        mean.append(np.mean(accs)); lo.append(np.percentile(accs, 2.5)); hi.append(np.percentile(accs, 97.5))
    return np.array(mean), np.array(lo), np.array(hi)

secs = [SEG_SEC + (K - 1) * SEG_SEC * (1 - OVERLAP) for K in CFG['dur_Ks']]
fig, axes = plt.subplots(1, 3, figsize=(16, 4), sharey=True); dur_rows = []
for ax, which in zip(axes, ['rest', 'photic', 'both']):
    m, lo, hi = duration_curve(oof[:, 0], which, np.random.default_rng(3))
    ax.plot(secs, m, marker='o', lw=2, label='NN fusion'); ax.fill_between(secs, lo, hi, alpha=.15)
    dur_rows += [dict(input=which, seconds=s_, bal_acc=a) for s_, a in zip(secs, m)]
    ax.axhline(1/3, color='gray', ls='--'); ax.set_xscale('log'); ax.set_title(f"input: {which}")
    ax.set_xlabel('EEG used per recording (s, random window)'); ax.grid(alpha=.3); ax.set_ylim(0, 1)
axes[0].set_ylabel('Subject-level balanced accuracy'); axes[0].legend()
plt.suptitle('Stage 6 - recording-duration analysis (out-of-fold, fusion head)'); plt.tight_layout()
plt.savefig(f"{OUT_DIR}/stage6_duration_curve.png", dpi=150); plt.show()
pd.DataFrame(dur_rows).to_csv(f"{OUT_DIR}/stage6_duration_curve.csv", index=False)

# ------------------------------------------------------------------
# 12. PAIRED COMPARISONS between heads (paired bootstrap on balanced accuracy + exact McNemar)
# ------------------------------------------------------------------
def paired_compare(yt, pa, pb, rng):
    """pa, pb: predicted labels of models A and B on the SAME subjects. Positive diff = A better."""
    n = len(yt); ca, cb = (pa == yt), (pb == yt)
    d_obs = balanced_accuracy_score(yt, pa) - balanced_accuracy_score(yt, pb)
    diffs = np.empty(CFG['n_boot'])
    for b in range(CFG['n_boot']):
        i = rng.integers(0, n, n); diffs[b] = balanced_accuracy_score(yt[i], pa[i]) - balanced_accuracy_score(yt[i], pb[i])
    lo, hi_ = np.percentile(diffs, [2.5, 97.5])
    p_boot = max(2 * min((diffs <= 0).mean(), (diffs >= 0).mean()), 1.0 / CFG['n_boot']); p_boot = min(p_boot, 1.0)
    b_, c_ = int((ca & ~cb).sum()), int((~ca & cb).sum())                      # discordant pairs
    p_mc = binomtest(b_, b_ + c_, 0.5).pvalue if (b_ + c_) > 0 else 1.0
    return dict(diff=d_obs, ci_lo=lo, ci_hi=hi_, p_boot=min(p_boot, 1.0), A_only_correct=b_, B_only_correct=c_, p_mcnemar=p_mc)

if CFG['run_pairs']:
    prng = np.random.default_rng(21)
    pred_e = {'fusion (headline)': pp_e.argmax(1), 'temporal': sc_t[ok_e].argmax(1), 'graph': sc_g[ok_e].argmax(1),
              'all-3 heads': NN_CANDS['NN all-3 heads'].argmax(1), 'classical ens.': CLS_ENS[ok_e].argmax(1)}
    PAIRS = [('fusion (headline)', 'temporal'), ('fusion (headline)', 'graph'), ('fusion (headline)', 'all-3 heads'),
             ('fusion (headline)', 'classical ens.'), ('graph', 'temporal'), ('all-3 heads', 'classical ens.')]
    pr_rows = []
    for a, b in PAIRS:
        pr_rows.append(dict(A=a, B=b, bal_acc_A=balanced_accuracy_score(yt_e_all, pred_e[a]),
                            bal_acc_B=balanced_accuracy_score(yt_e_all, pred_e[b]), **paired_compare(yt_e_all, pred_e[a], pred_e[b], prng)))
    pr_df = pd.DataFrame(pr_rows); pr_df['q_mcnemar_BH'] = bh(pr_df.p_mcnemar.values)
    pr_df.to_csv(f"{OUT_DIR}/paired_comparisons_heads.csv", index=False)
    print("\n" + "=" * 78 + "\nPAIRED COMPARISONS (rest+photic, same subjects; diff = A - B; McNemar exact; BH over this table)\n" + "=" * 78)
    print(pr_df.round(3).to_string(index=False))
    print("Note: with n~88 subjects, differences of a few points are within noise -- say so rather than claiming a 'best' head.")

# ------------------------------------------------------------------
# 13. LENGTH CONFOUND: evaluate on a FIXED number of segments per recording
# ------------------------------------------------------------------
def make_matched_groups(pct):
    cnts = np.array([len(i) for i in D.groups.values()])
    L = int(max(MIN_SEGS, np.percentile(cnts, pct))); G = {}
    for k, idx in D.groups.items():
        if len(idx) > L: st = (len(idx) - L) // 2; G[k] = idx[st:st + L]       # centre window (label-free)
        else: G[k] = idx
    return L, G

if CFG['run_lenmatch']:
    L_FIX, GM = make_matched_groups(CFG['len_match_pct'])
    n_trunc = sum(len(D.groups[k]) > L_FIX for k in D.groups)
    print("\n" + "=" * 78 + f"\nLENGTH-MATCHED EVALUATION: {L_FIX} segments (~{SEG_SEC + (L_FIX-1)*SEG_SEC*(1-OVERLAP):.0f} s) per recording; "
          f"{n_trunc}/{len(D.groups)} recordings truncated, the rest are shorter and kept as is\n" + "=" * 78)
    len_tab = pd.DataFrame([dict(group=CLASS_NAMES[D.subj_y[s]], cond=COND_NAMES[c], n_full=len(D.groups[(s, c)]), n_matched=len(GM[(s, c)]))
                            for (s, c) in D.groups])
    print("mean #segments per recording (full -> matched):\n", len_tab.groupby(['cond', 'group'])[['n_full', 'n_matched']].mean().round(1).to_string())
    RAW_COUNTS_M = np.zeros((S, 2))
    for (s, c), idx in GM.items(): RAW_COUNTS_M[s, c] = len(idx)
    X_LEN_M = np.log1p(RAW_COUNTS_M).astype(np.float32)
    X_REG_M, X_FULL_M = build_feature_matrices(GM)

    lm_rows = []
    m_, s_, P0 = classical_cv(X_LEN, LR_LEN); m2, s2, _ = classical_cv(X_LEN_M, LR_LEN)
    lm_rows.append(dict(model='LENGTH_ONLY probe', full=m_, matched=m2))
    P_m = {}
    for nm in CLS_MEMBERS:
        m1, _, _ = classical_cv(X_REG, MODELS[nm]); m2, _, P_m[nm] = classical_cv(X_REG_M, MODELS[nm])
        lm_rows.append(dict(model=nm, full=m1, matched=m2))
    m1, _, _ = classical_cv(X_FULL, LR_FULL); m2, _, _ = classical_cv(X_FULL_M, LR_FULL)
    lm_rows.append(dict(model='LR_fullfeat', full=m1, matched=m2))
    ENS_M = np.mean([P_m[k] for k in CLS_MEMBERS], 0)
    lm_rows.append(dict(model='CLASSICAL_ENSEMBLE (fold0)', full=balanced_accuracy_score(D.subj_y, CLS_ENS.argmax(1)),
                        matched=balanced_accuracy_score(D.subj_y, ENS_M.argmax(1))))
    lrng = np.random.default_rng(5)
    for hi_i, head in enumerate(HEADS):
        for which in ['rest', 'photic', 'both']:
            sc_m = subject_scores(oof[:, hi_i], which, groups=GM); ok_m = ~np.isnan(sc_m).any(1)
            sm = summarize(D.subj_y[ok_m], softmax_np(sc_m[ok_m]), lrng)
            lm_rows.append(dict(model=f"NN {head} [{which}]", full=res[(res['head'] == head) & (res['input'] == which)].bal_acc.values[0],
                                matched=sm['bal_acc'], matched_ci_lo=sm['ci_lo'], matched_ci_hi=sm['ci_hi']))
    lm_df = pd.DataFrame(lm_rows); lm_df.to_csv(f"{OUT_DIR}/length_matched_results.csv", index=False)
    print(lm_df.round(3).to_string(index=False))
    print("Reading guide: LENGTH_ONLY should drop toward 0.333 after matching; if the NN / classical numbers stay roughly "
          "unchanged, recording length is not what drives them. Training already caps each recording at "
          f"{CFG['sup_segs_per_rec']} segments, so the NN never sees length as a feature. State this in the paper.")

# ------------------------------------------------------------------
# 14. DEMOGRAPHIC CONFOUNDS (age / sex / MMSE)
# ------------------------------------------------------------------
def load_demographics():
    if SMOKE_TEST:
        r = np.random.default_rng(1); n = S
        return pd.DataFrame(dict(age=r.integers(55, 80, n).astype(float), sex=r.choice(['M', 'F'], n), mmse=r.integers(10, 31, n).astype(float)))
    df = pd.read_csv(os.path.join(PATHS['rest'], 'participants.tsv'), sep='\t')
    df.columns = [c.strip().lower() for c in df.columns]
    df['participant_id'] = df['participant_id'].astype(str).str.strip()
    df = df.drop_duplicates('participant_id').set_index('participant_id').reindex(D.sub_ids)
    acol = next((c for c in df.columns if c == 'age'), None)
    scol = next((c for c in df.columns if c in ('gender', 'sex')), None)
    mcol = next((c for c in df.columns if 'mmse' in c), None)
    out = pd.DataFrame(index=range(S))
    out['age'] = pd.to_numeric(df[acol], errors='coerce').values if acol else np.nan
    if scol:
        sx = df[scol].astype(str).str.strip().str.upper().str[0]; out['sex'] = sx.where(sx.isin(['M', 'F'])).values
    else: out['sex'] = np.nan
    out['mmse'] = pd.to_numeric(df[mcol], errors='coerce').values if mcol else np.nan
    return out

if CFG['run_demo']:
    print("\n" + "=" * 78 + "\nDEMOGRAPHICS BY GROUP\n" + "=" * 78)
    try:
        dem = load_demographics(); dem['label'] = [CLASS_NAMES[i] for i in D.subj_y]
        dem['female'] = np.where(dem.sex.isna(), np.nan, (dem.sex == 'F').astype(float))
        tab = []
        for k in CLASS_NAMES:
            g = dem[dem.label == k]
            tab.append(dict(group=k, n=len(g), age_mean_sd=f"{g.age.mean():.1f} +/- {g.age.std():.1f}",
                            age_range=f"{g.age.min():.0f}-{g.age.max():.0f}", female_n=int(g.female.sum()),
                            female_pct=f"{100*g.female.mean():.0f}%", mmse_mean_sd=f"{g.mmse.mean():.1f} +/- {g.mmse.std():.1f}"))
        tab = pd.DataFrame(tab); print(tab.to_string(index=False)); tab.to_csv(f"{OUT_DIR}/demographics_by_group.csv", index=False)
        grp = lambda col: [dem[dem.label == k][col].dropna().values for k in CLASS_NAMES]
        if all(len(g) > 1 for g in grp('age')): print(f"Age   Kruskal-Wallis across groups: p = {kruskal(*grp('age'))[1]:.4f}")
        if all(len(g) > 1 for g in grp('mmse')): print(f"MMSE  Kruskal-Wallis across groups: p = {kruskal(*grp('mmse'))[1]:.4g}  (MMSE is part of the diagnosis -> expected to differ)")
        ct = pd.crosstab(dem.label, dem.sex)
        if ct.shape[0] >= 2 and ct.shape[1] >= 2: print(f"Sex   chi-square across groups:    p = {chi2_contingency(ct)[1]:.4f}")
        for a, b in [('AD', 'CN'), ('FTD', 'CN'), ('AD', 'FTD')]:
            xa, xb = dem[dem.label == a].age.dropna(), dem[dem.label == b].age.dropna()
            if len(xa) > 2 and len(xb) > 2: print(f"  age {a} vs {b}: Mann-Whitney p = {mannwhitneyu(xa, xb)[1]:.4f}")

        # (a) demographics-only baselines (same subject-level CV)
        X_DEM = np.c_[dem.age.values, dem.female.values].astype(np.float32)
        X_DEMM = np.c_[X_DEM, dem.mmse.values].astype(np.float32)
        b1 = classical_cv(X_DEM, LR_LEN)[:2]; b2 = classical_cv(X_DEMM, LR_LEN)[:2]
        print(f"\nDemographics-only classifier (age + sex):         balanced acc {b1[0]:.3f} +/- {b1[1]:.3f}  (chance 0.333)")
        print(f"Demographics + MMSE (diagnostic info, reference): balanced acc {b2[0]:.3f} +/- {b2[1]:.3f}")

        # (b) does model correctness depend on age / sex?   (headline model, out-of-fold)
        dd = dem.iloc[np.where(ok_e)[0]].reset_index(drop=True); dd['correct'] = (pp_e.argmax(1) == yt_e_all)
        ca, ic = dd[dd.correct].age.dropna(), dd[~dd.correct].age.dropna()
        if len(ca) > 2 and len(ic) > 2:
            print(f"\nHeadline correct vs incorrect subjects: age {ca.mean():.1f} vs {ic.mean():.1f} (Mann-Whitney p = {mannwhitneyu(ca, ic)[1]:.3f})")
        ct2 = pd.crosstab(dd.correct, dd.sex)
        if ct2.shape == (2, 2): print(f"  sex vs correctness: Fisher exact p = {fisher_exact(ct2)[1]:.3f}")
        print("  accuracy by sex:\n", dd.groupby('sex').correct.agg(['mean', 'count']).round(3).to_string())
        try:
            dd['age_tertile'] = pd.qcut(dd.age, 3, duplicates='drop')
            print("  accuracy by age tertile:\n", dd.groupby('age_tertile').correct.agg(['mean', 'count']).round(3).to_string())
        except Exception: pass
        print("Reading guide: if age+sex alone are near chance and accuracy is similar across age/sex strata, demographics "
              "are unlikely to drive the result. Report the table above in the paper (dataset description).")
        tab_out = dd.groupby('sex').correct.agg(['mean', 'count']).reset_index(); tab_out.to_csv(f"{OUT_DIR}/accuracy_by_sex.csv", index=False)
    except Exception as e:
        print(f"Demographic analysis skipped: {type(e).__name__}: {e}  (check participants.tsv columns: age / gender / mmse)")

# ------------------------------------------------------------------
# 15. LABEL-PERMUTATION TEST (subject labels shuffled, folds unchanged)
# ------------------------------------------------------------------
if CFG['run_perm']:
    print("\n" + "=" * 78 + "\nLABEL-PERMUTATION TEST\n" + "=" * 78)
    obs_c = balanced_accuracy_score(D.subj_y, cls_P['LR_regional'].argmax(1))
    prng_ = np.random.default_rng(99); null_c = []
    for p_ in range(CFG['clf_perm']):
        yp = prng_.permutation(D.subj_y)
        _, _, Pp = classical_cv(X_REG, MODELS['LR_regional'], y=yp, reps=1)
        null_c.append(balanced_accuracy_score(yp, Pp.argmax(1)))
    null_c = np.array(null_c); p_c = (1 + (null_c >= obs_c).sum()) / (1 + len(null_c))
    print(f"Classical LR_regional : observed {obs_c:.3f} | null mean {null_c.mean():.3f} (sd {null_c.std():.3f}, max {null_c.max():.3f}) "
          f"| permutation p = {p_c:.4f} ({len(null_c)} perms)")
    nn_null = []
    for pi in range(CFG['nn_perm']):
        yp_subj = np.random.default_rng(1000 + pi).permutation(D.subj_y).astype(np.int32); yp_seg = yp_subj[D.subj]
        o_p, fl_p, _ = run_nn_cv(f"PERM{pi+1}", yp_subj, yp_seg, BASE_VAR, CFG['perm_seeds'], verbose=False)
        sc_p = subject_scores(o_p.mean(0)[:, 0], 'both', fl=fl_p); ok_p = ~np.isnan(sc_p).any(1)
        nn_null.append(balanced_accuracy_score(yp_subj[ok_p], sc_p[ok_p].argmax(1)))
        print(f"NN fusion, permuted labels run {pi+1}/{CFG['nn_perm']}: balanced acc = {nn_null[-1]:.3f}")
    if nn_null:
        print(f"NN fusion  : observed headline {hl['bal_acc']:.3f} | permuted-label runs mean {np.mean(nn_null):.3f} "
              f"(min {np.min(nn_null):.3f}, max {np.max(nn_null):.3f}) -> should sit near 0.333; "
              f"with {len(nn_null)} runs this is a sanity check, not a p-value (use the classical p for that).")
    pd.DataFrame(dict(kind=['classical'] * len(null_c) + ['nn_fusion'] * len(nn_null), bal_acc=list(null_c) + list(nn_null))
                 ).to_csv(f"{OUT_DIR}/permutation_null.csv", index=False)
    fig, ax = plt.subplots(figsize=(6, 4))
    ax.hist(null_c, bins=20, alpha=.7, label='classical LR, permuted labels')
    ax.axvline(obs_c, color='r', label=f'observed LR_regional ({obs_c:.2f})'); ax.axvline(1/3, color='gray', ls='--')
    if nn_null: ax.scatter(nn_null, np.zeros(len(nn_null)), marker='x', c='k', s=50, label='NN fusion, permuted labels', zorder=5)
    ax.axvline(hl['bal_acc'], color='g', ls=':', label=f"NN headline ({hl['bal_acc']:.2f})")
    ax.set_xlabel('Subject-level balanced accuracy'); ax.set_ylabel('count'); ax.legend(fontsize=8); ax.grid(alpha=.3)
    plt.tight_layout(); plt.savefig(f"{OUT_DIR}/permutation_test.png", dpi=150); plt.show()

# ------------------------------------------------------------------
# 16. ABLATIONS of the v6 changes (each on the same folds / seeds, paired against the full v6 model)
# ------------------------------------------------------------------
ABL = {
    '- modality dropout':                       dict(mod_drop=0.0),
    '- train+val (early-stop on val, 56 subj)': dict(trainval=False),
    '- multi-scale conv (single 0.25 s kernel)': dict(tmp_kernels=(32,), tmp_f_per_k=3 * TMP_F_PER_K),
    '- head weights (graph 0.7 -> 0.3)':        dict(head_w=(1.0, 0.7, 0.3)),
    '- ALL FOUR (v5-like)':                     dict(mod_drop=0.0, trainval=False, tmp_kernels=(32,), tmp_f_per_k=3 * TMP_F_PER_K, head_w=(1.0, 0.7, 0.3)),
}
if CFG['run_ablations']:
    k_ab = min(CFG['abl_seeds'], CFG['n_seeds'])
    print("\n" + "=" * 78 + f"\nABLATIONS ({k_ab} NN seeds each; reference = first {k_ab} seeds of the main run; positive delta = component helps)\n" + "=" * 78)
    def three_scores(o):
        f, t, g = [subject_scores(o[:, h], 'both', fl=filled_ab) for h in range(3)]
        return f, t, g
    filled_ab = filled
    ref = oof_s[:k_ab].mean(0); r_f, r_t, r_g = three_scores(ref)
    arng = np.random.default_rng(31); ab_rows = []
    ok_r = ~np.isnan(r_f).any(1) & ~np.isnan(r_t).any(1) & ~np.isnan(r_g).any(1)
    ab_rows.append(dict(variant='FULL v6 (reference)', fusion=balanced_accuracy_score(D.subj_y[ok_r], r_f[ok_r].argmax(1)),
                        all3=balanced_accuracy_score(D.subj_y[ok_r], ((r_f + r_t + r_g)[ok_r]).argmax(1)),
                        graph_only=balanced_accuracy_score(D.subj_y[ok_r], r_g[ok_r].argmax(1)),
                        temporal_only=balanced_accuracy_score(D.subj_y[ok_r], r_t[ok_r].argmax(1))))
    for name, ch in ABL.items():
        var = {**BASE_VAR, **ch}
        o_a, fl_a, _ = run_nn_cv('ABL ' + name, D.subj_y, D.y, var, k_ab, verbose=False)
        filled_ab = fl_a; a_f, a_t, a_g = three_scores(o_a.mean(0)); filled_ab = filled
        ok = ok_r & ~np.isnan(a_f).any(1) & ~np.isnan(a_t).any(1) & ~np.isnan(a_g).any(1); yt = D.subj_y[ok]
        cmp_f = paired_compare(yt, r_f[ok].argmax(1), a_f[ok].argmax(1), arng)       # reference minus ablated (fusion head)
        ab_rows.append(dict(variant=name, fusion=balanced_accuracy_score(yt, a_f[ok].argmax(1)),
                            all3=balanced_accuracy_score(yt, (a_f + a_t + a_g)[ok].argmax(1)),
                            graph_only=balanced_accuracy_score(yt, a_g[ok].argmax(1)),
                            temporal_only=balanced_accuracy_score(yt, a_t[ok].argmax(1)),
                            delta_fusion_ref_minus_abl=cmp_f['diff'], delta_ci_lo=cmp_f['ci_lo'], delta_ci_hi=cmp_f['ci_hi'],
                            p_boot=cmp_f['p_boot'], p_mcnemar=cmp_f['p_mcnemar']))
        print(f"  done: {name}: fusion {ab_rows[-1]['fusion']:.3f}  (delta {cmp_f['diff']:+.3f}, McNemar p = {cmp_f['p_mcnemar']:.3f})")
    ab_df = pd.DataFrame(ab_rows)
    if 'p_mcnemar' in ab_df: ab_df['q_mcnemar_BH'] = np.r_[np.nan, bh(ab_df.p_mcnemar.values[1:])]
    ab_df.to_csv(f"{OUT_DIR}/ablation_results.csv", index=False)
    print(ab_df.round(3).to_string(index=False))
    print("Reading guide: only claim a component 'helps' if delta > 0 AND its CI excludes 0 (or q < 0.05). "
          "Otherwise write 'no measurable effect at n=88 subjects' -- that is an honest and publishable statement.")

# ------------------------------------------------------------------
# 17. LIMITATIONS (paste/adapt into the paper)
# ------------------------------------------------------------------
LIMITS = f"""LIMITATIONS (suggested wording)
1. Single cohort, no external validation: all 88 subjects come from one public dataset (ds004504, with the photic
   recordings from ds006036). Performance may not transfer to other scanners, montages, protocols or populations.
2. Diagnostic separation, not early detection: the design is cross-sectional (AD vs FTD vs CN at one time point).
   Results show how well the three diagnoses can be separated, NOT how early disease can be detected.
   The recording-duration analysis answers 'how much EEG is needed', not 'how early'.
3. Needs both recordings: the best results use eyes-closed rest AND photic recordings per subject; single-condition
   accuracy is lower (see subject_level_results.csv).
4. Small sample: n = 88 subjects; confidence intervals are wide and head-to-head differences of a few points are not
   significant (see paired_comparisons_heads.csv).
5. Researcher degrees of freedom: architecture/training choices (v2-v6) were developed on these same subjects. v7 uses a
   new seed (CFG seed = {CFG['seed']}) and a pre-declared headline, but not new subjects, so estimates remain somewhat optimistic.
6. Recording length differs between groups (LENGTH_ONLY probe above chance); results were re-evaluated on length-matched
   recordings (length_matched_results.csv). Age/sex/MMSE were checked but diagnosis-related confounds cannot be fully excluded.
"""
print("\n" + "=" * 78 + "\n" + LIMITS + "=" * 78)
with open(f"{OUT_DIR}/limitations_statement.txt", "w") as f_: f_.write(LIMITS)

print("\nDONE. Files in", OUT_DIR, ": classical_results.csv, subject_level_results.csv, ensemble_results.csv, "
      "paired_comparisons_heads.csv, length_matched_results.csv, demographics_by_group.csv, permutation_null.csv, "
      "ablation_results.csv, limitations_statement.txt, oof_predictions.npz, cv_results.png, confusion_subject_level.png, "
      "permutation_test.png, stage4_*, stage5_*, stage6_*")
```
